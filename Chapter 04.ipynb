## 4.1 Metrik Jarak & Algoritma KNN
Perhitungan jarak Euclidean, Manhattan, dan Minkowski pada model instance-based.

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report

X, y = load_wine(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

scaler = StandardScaler()
X_tr_sc = scaler.fit_transform(X_tr)
X_te_sc = scaler.transform(X_te)
```

## 4.2 Tuning Hyperparameter KNN via GridSearchCV

```python
param_grid = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan']
}

grid_knn = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5, scoring='accuracy')
grid_knn.fit(X_tr_sc, y_tr)

print("Hyperparameter KNN Terbaik:", grid_knn.best_params_)
print(f"Akurasi Validasi Silang Terbaik: {grid_knn.best_score_*100:.2f}%")
print("
Laporan Evaluasi Klasifikasi:
", classification_report(y_te, grid_knn.best_estimator_.predict(X_te_sc)))
```
