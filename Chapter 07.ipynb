## 7.1 Support Vector Classifier (SVC) & Kernel Tuning

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

X, y = load_iris(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42)

param_grid = {
    'C': [0.1, 1, 10],
    'gamma': ['scale', 'auto', 0.1],
    'kernel': ['linear', 'rbf', 'poly']
}

grid_svc = GridSearchCV(SVC(), param_grid, cv=5).fit(X_tr, y_tr)
print("Parameter SVC Terbaik:", grid_svc.best_params_)
print("Akurasi Pengujian    :", accuracy_score(y_te, grid_svc.best_estimator_.predict(X_te)))
```
