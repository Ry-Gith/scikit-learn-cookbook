## 2.1 Imputasi Nilai Hilang (Missing Values)
Mengatasi NaN menggunakan `SimpleImputer` (Median/Modus) dan `KNNImputer` (Weighted Distance).

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.preprocessing import StandardScaler, RobustScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

df_raw = pd.DataFrame({
    'Umur': [22, 28, np.nan, 42, 35, 55, 110],
    'Gaji': [4500000, 6500000, 5800000, np.nan, 11000000, 18000000, 24000000],
    'Kota': ['Jakarta', 'Bandung', 'Jakarta', 'Surabaya', np.nan, 'Bandung', 'Jakarta']
})
print("Data Mentah:
", df_raw)
```

## 2.2 Feature Scaling (StandardScaler vs RobustScaler)
Membandingkan penskalaan Z-Score dengan RobustScaler yang tahan pencilan.

```python
num_cols = ['Umur', 'Gaji']
imputed_num = SimpleImputer(strategy='median').fit_transform(df_raw[num_cols])

std_scaled = StandardScaler().fit_transform(imputed_num)
rob_scaled = RobustScaler().fit_transform(imputed_num)

print("Output StandardScaler:
", np.round(std_scaled[:3], 3))
print("Output RobustScaler:
", np.round(rob_scaled[:3], 3))
```

## 2.3 ColumnTransformer untuk Dataset Heterogen
Transformasi terpisah untuk variabel numerik dan kategorikal.

```python
num_pipeline = Pipeline([
    ('imputer', KNNImputer(n_neighbors=2)),
    ('scaler', RobustScaler())
])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(sparse_output=False, handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', num_pipeline, ['Umur', 'Gaji']),
    ('cat', cat_pipeline, ['Kota'])
])

X_transformed = preprocessor.fit_transform(df_raw)
print("Bentuk Matriks Preprocessing Akhir:", X_transformed.shape)
print("Sampel Matriks Preprocessing:
", np.round(X_transformed[:4], 2))
```
