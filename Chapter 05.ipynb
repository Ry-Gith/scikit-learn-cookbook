## 5.1 Regularisasi Regresi Linier (Ridge, Lasso, ElasticNet)

```python
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.preprocessing import SplineTransformer
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error, r2_score

np.random.seed(42)
X = np.sort(np.random.rand(100, 1) * 6 - 3, axis=0)
y = np.sin(X).ravel() + np.random.normal(0, 0.1, 100)

models = {
    'OLS': LinearRegression(),
    'Ridge (L2)': Ridge(alpha=1.0),
    'Lasso (L1)': Lasso(alpha=0.05),
    'ElasticNet': ElasticNet(alpha=0.05, l1_ratio=0.5)
}

for name, model in models.items():
    model.fit(X, y)
    preds = model.predict(X)
    print(f"{name:12s} | MSE: {mean_squared_error(y, preds):.4f} | R2: {r2_score(y, preds):.4f}")
```

## 5.2 Spline Regression untuk Kurva Non-Linier

```python
spline_model = make_pipeline(SplineTransformer(n_knots=5, degree=3), Ridge(alpha=0.1))
spline_model.fit(X, y)
spline_preds = spline_model.predict(X)

print("Spline Ridge MSE:", mean_squared_error(y, spline_preds))
print("Spline Ridge R2 :", r2_score(y, spline_preds))
```
