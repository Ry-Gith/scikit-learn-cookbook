## 13.1 Serialisasi Model (`joblib`) & Production Verification

```python
import joblib, os
import numpy as np
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

X, y = load_iris(return_X_y=True)
clf = RandomForestClassifier(random_state=42).fit(X, y)

joblib.dump(clf, 'model.joblib')
loaded_clf = joblib.load('model.joblib')

assert np.array_equal(clf.predict(X[:5]), loaded_clf.predict(X[:5]))
print("Verifikasi Serialisasi Model BERHASIL! Prediksi 100% identik.")

if os.path.exists('model.joblib'):
    os.remove('model.joblib')
```
