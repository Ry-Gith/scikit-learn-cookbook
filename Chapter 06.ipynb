## 6.1 Regresi Logistik Biner & Analisis ROC-AUC

```python
from sklearn.datasets import load_breast_cancer, make_multilabel_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.multioutput import MultiOutputClassifier
from sklearn.metrics import roc_auc_score

X, y = load_breast_cancer(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

clf = LogisticRegression(max_iter=5000, solver='liblinear').fit(X_tr, y_tr)
proba = clf.predict_proba(X_te)[:, 1]
print("Skor ROC-AUC Biner:", roc_auc_score(y_te, proba))
```

## 6.2 Multilabel Classification (`MultiOutputClassifier`)

```python
X_ml, y_ml = make_multilabel_classification(n_samples=200, n_features=10, n_classes=3, random_state=42)
multi_clf = MultiOutputClassifier(LogisticRegression(solver='liblinear')).fit(X_ml, y_ml)
print("Prediksi Multilabel 3 Sampel Pertama:
", multi_clf.predict(X_ml[:3]))
```
