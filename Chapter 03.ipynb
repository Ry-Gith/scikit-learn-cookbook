## 3.1 Principal Component Analysis (PCA)
Memetakan fitur berdimensi tinggi ke komponen ortogonal (Unsupervised).

```python
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.manifold import TSNE
import numpy as np

digits = load_digits()
X, y = digits.data, digits.target
X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=10)
X_pca = pca.fit_transform(X_scaled)

print("Explained Variance Ratio per Komponen:
", np.round(pca.explained_variance_ratio_, 3))
print(f"Total Varians Terjelaskan (10 Komponen): {np.sum(pca.explained_variance_ratio_)*100:.2f}%")
```

## 3.2 Linear Discriminant Analysis (LDA)
Memaksimalkan varians antar-kelas dan meminimalkan varians di dalam kelas (Supervised).

```python
lda = LinearDiscriminantAnalysis(n_components=2)
X_lda = lda.fit_transform(X_scaled, y)
print("Bentuk Matriks Proyeksi LDA:", X_lda.shape)
print("Explained Variance Ratio LDA:", np.round(lda.explained_variance_ratio_, 3))
```

## 3.3 t-SNE Manifold Learning
Visualisasi non-linier untuk struktur klaster berdimensi tinggi.

```python
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
X_tsne = tsne.fit_transform(X_scaled[:400])
print("Bentuk Output t-SNE (400 sampel):", X_tsne.shape)
```
