## 8.1 Random Forest vs Gradient Boosting & Feature Importances

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

cancer = load_breast_cancer()
X, y = cancer.data, cancer.target
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42)

rf = RandomForestClassifier(n_estimators=100, random_state=42).fit(X_tr, y_tr)
gb = GradientBoostingClassifier(n_estimators=100, random_state=42).fit(X_tr, y_tr)

print(f"Random Forest Acc    : {accuracy_score(y_te, rf.predict(X_te))*100:.2f}%")
print(f"Gradient Boosting Acc: {accuracy_score(y_te, gb.predict(X_te))*100:.2f}%")

imp = pd.Series(rf.feature_importances_, index=cancer.feature_names).sort_values(ascending=False)
print("
Top 5 Fitur Terpenting (Random Forest):
", imp.head(5))
```
